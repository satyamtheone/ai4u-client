"""
    This notebook is being used to generate orders data. I'm aiming to generator 10,000 data points in total
    having 500 as RTO cases. 
    
    Different logic are being used - 
    1. random rumber in range - order_no, amount, item-quantity, 
    2. random generator for "email_id", "mobile_no", timestamp
    3. different publicly available data set is being used sampling - Name, address, pincode, product_categories 
    4. random boolean value for first time order 

"""

```python
import pandas as pd
import math
import numpy as np
import random
import datetime
import string
```

## Data generation

```python
c = 9500
total_rto = 1000
```

```python
features_init = ["order_no", "amount", "item_quantity", "product_category", "address",
                 "customer_name", "email_id", "mobile_no", "pincode", "first_time_order", "timestamp", "RTO"] 
len(features_init)
```

#### load indian name and address

```python
df_name_address = pd.read_csv("./RTO_files/indian_name_address_data.csv")
df_name_address.head()
```

#### load indian pincode

```python
df_pincodes = pd.read_csv("./RTO_files/all_india_PO_list_without_APS_offices_ver2_lat_long.csv")
df_pincodes.head()
```

#### load product categories

```python
df_prod_categories = pd.read_csv("./RTO_files/product_categories.csv")
df_prod_categories.head()
```

```python
#choose random prod categories from df_prod_categories
def get_prod_categories(total):
    for count in range(total):
        yield random.choice(list(df_prod_categories["prod_categ"]))
    
```

```python
#random date generator
#here i'm generating for 1 year back from current date
def get_random_date_list(total_non_rto):
    #list_len is total_non_rto
    for x_ran in random.sample(range(10, 9999), total_non_rto):
     yield  str(datetime.datetime.now() + datetime.timedelta(minutes=random.randrange(x_ran)) - datetime.timedelta(weeks=52))
    
def mobile_no_generator(total, valid_bool):
    for times in range(total):
        mobile_no = ""
    
        #first number between 6-9
        mobile_no = mobile_no + str(random.randint(6, 9))
        for i in range(1, 10):
            mobile_no = mobile_no + str(random.randint(0, 9))
            
        if valid_bool:
            yield mobile_no
        else:
            mobile_no = mobile_no + str(random.randint(0, 9))
            if random.random()<.5:
                yield mobile_no
            else:
                yield mobile_no[:9]
            
            

        
def email_id_generator(total, valid_bool):
    for times in range(total):
        char_num = random.randint(4,10)
        email_char =  ''.join(random.choice(string.ascii_letters) for _ in range(char_num))
        
        
        if valid_bool:
            email_id = email_char+ random.choice(["@gmail.com", "@bing.com", "@reddit.com"])
            yield email_id
        else:
            email_id = email_char+ random.choice(["@gmailcom", "com", "@.reddit"])
            yield email_id
```

```python
df_orders = pd.DataFrame({
    "order_no" : random.sample(range(1000000, 9999999), total_non_rto),
    "amount" : np.random.randint(50,10000, size=total_non_rto),
    "item_quantity": np.random.randint(1,20, size=total_non_rto),
    "product_category": get_prod_categories(total_non_rto),
    "address": random.sample(list(df_name_address["address"]), total_non_rto),
    "customer_name": random.sample(list(df_name_address["Name"]), total_non_rto),
    "email_id": email_id_generator(total_non_rto, True),
    "mobile_no": mobile_no_generator(total_non_rto, True),
    "pincode": random.sample(list(df_pincodes["pincode"]), total_non_rto),
    "first_time_order": np.random.randint(0,2, size=total_non_rto),
    "timestamp": get_random_date_list(total_non_rto),
    "RTO": [0 if random.random()<.98 else 1 for _ in range(total_non_rto)]
})
df_orders.head()
```

```python
## loading gibberish data for address
df_gibberish = pd.read_csv("./RTO_files/gibberish.csv")
df_gibberish.head()
```

```python
## functions to have missing names, not valid address, or irregular pincodes

def get_names_rto(total_rto):
    for _ in range(total_rto):
        if random.random()>.5:
            yield random.choice(list(df_name_address["Name"]))
        else:
            yield ""

def get_address_rto(total_rto):
    for _ in range(total_rto):
        if random.random()>.7:
            yield random.choice(list(df_name_address["address"]))
        elif random.random()>.5:
            yield ""
        else:
            #gibberish
            yield random.choice(list(df_gibberish["gibberish"]))

def get_pincode_rto(total_rto):  
    for _ in range(total_rto):
        if random.random()>.5:
            yield random.choice(list(df_pincodes["pincode"]))
        else:
            yield random.randint(1000000, 99999999)
        
            
```

```python
df_orders_RTO = pd.DataFrame({
    "order_no" : random.sample(range(10000000, 99999999), total_rto),
    "amount" : np.random.randint(50,100000, size=total_rto),
    "item_quantity": np.random.randint(1,2000, size=total_rto),
    "product_category": get_prod_categories(total_rto),
    "address": get_address_rto(total_rto),
    "customer_name": get_names_rto(total_rto),
    "email_id": email_id_generator(total_rto, False),
    "mobile_no": mobile_no_generator(total_rto, False),
    "pincode": get_pincode_rto(total_rto),
    "first_time_order": np.random.randint(0,2, size=total_rto),
    "timestamp": get_random_date_list(total_rto),
    "RTO":1
        
})
df_orders_RTO.head()
```
