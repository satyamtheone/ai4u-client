## Problem formulation
RTO are quite prevalent in e-commerce. Some of the RTO are justified (wrong item, defective items etc) but fraudulent cases do impact business negatively. Given the scale and velocity of orders in e-commerce, an ML algorithm to identify one is the right approach. 
Data available for total RTO will always be imbalanced compared to total orders. Since we are trying to identify RTO or not, it could be  formulated into a binary classification problem.  
I have generated dummy data with following features - "order_no", "amount", "item_quantity", "product_category", "address",  "customer_name", "email_id", "mobile_no", "pincode", "first_time_order", "timestamp". Please check the Data_generation file to see the logic for each feature generated. 
There is a target column “RTO” having value 0 (successful delivery) and 1 (RTO). 
 
 
## Approach
Since this is a case of classification on imbalanced data we can take below approaches:
1. sampling technique (over sampling, undersampling) to create balance of both class
2. SMOTE technique to increase positive class sample
3. Algorithm which works in imbalanced data case
 
We are going with the 3rd option here. I will be using xgboost with scale_pos_weight parameters. It internally handles imbalanced class cases. 
 
I will be using "precision" metrics. We could also use recall here depending on how this result will be used. If this is being used directly to cancel such an order, "precision" would be the right metrics as we don't want to affect genuine customer sentiments. If the result is being handled by a human agent to verify or cross-check with customers for missing information or just order confirmation then the recall metric would be right.
 
## Other approaches or next step
There are other possible enhancements:
1. Data collection:
    We could also collect data like ip's, device ids, age, customer_loyality etc. These would help model in making 
    better decisions
    
    
2. Feature engineering:
    Following features could have been created/calculated:
    1. clustering of pincode, age group into 2-3 groups like most frequent RTO, minimum RTO groups
    2. customer sign-up date could be used to calculate years of relationship.
    3. Analysis of text in address and Name like if it's monkey-typed, missing etc.
    
 
3. Algorithms:
    Other algorithms (like logistic regression, encode-decoder based Neural network etc) along with metrics like precision and recall could be trained. Even a ensemble of model could be used depending on performance of individual models

### feature engineering

Based on above analysis following features could be added:
    1. high amount transaction
    2. high quantity
    3. correct_mobile_no
    4. correct_email
    5. correct_pincode
    6. missing_critical_info   (missing name, address)
    
Further for later part we could add text verification to identify giberish name and address

```python
import re
```

```python
regex = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b' #"^[a-zA-Z0-9+_.-]+@[a-zA-Z0-9.-]+$"
len(re.findall(regex, "avinish@gmail.com"))
```

```python
orders_df["high_amount_trans"] = orders_df["amount"].apply(lambda x:1 if x>15000 else 0)
orders_df["item_quant_high"] = orders_df["item_quantity"].apply(lambda x:1 if x>19 else 0)
orders_df["correct_mobile_no"] = orders_df["mobile_no"].apply(lambda x:1 if len(str(x))==10 else 0)
orders_df["correct_pincode"] = orders_df["pincode"].apply(lambda x:1 if len(str(x))==6 else 0)
orders_df["missing_critical_info"] = pd.isna(orders_df[["customer_name", "address"]]).apply(lambda x: 1 \
                                                  if (x["customer_name"]+x["address"])>0 else 0, axis=1)

orders_df["correct_email"] = orders_df["correct_email"] = orders_df["email_id"].apply(lambda x: 1  \
                                                            if len(re.findall(regex,x))>0 else 0)
```

### data preparation for model building

we will go ahead with engineered features for model building.

```python
from sklearn.model_selection import train_test_split
```

```python
feature_select = ['high_amount_trans', 'item_quant_high', 'correct_mobile_no', 
                  'correct_pincode', 'missing_critical_info',  'correct_email']
```

```python
final_orders_features = orders_df[feature_select]
final_orders_y = orders_df["RTO"]


final_orders_features.head(2)

```

```python
X_train, X_test, y_train, y_test = train_test_split(final_orders_df, final_orders_y, train_size=.7, 
                                                    stratify=final_orders_y )
```

```python
#further breaking test into test and validation set
X_test, X_valid, y_test, y_valid = train_test_split(X_test, y_test, train_size=.5, 
                                                    stratify=y_test )
```

### model fitting

```python
import xgboost as xgb   ## !pip install xgboost
from xgboost import plot_importance, plot_tree
from sklearn.metrics import mean_squared_error, classification_report, precision_score
from sklearn.model_selection import GridSearchCV
import time
```

```python
parameters = {
    'n_estimators': [100, 250, 400],
    'learning_rate': [0.001, 0.1],
    'max_depth': [8, 15], 
    'gamma': [0.001, 0.01, 0.02],
    'scale_pos_weight': [1, 10, 50, 80 ,100]

}
```

```python
eval_set = [(X_train, y_train), (X_valid, y_valid)]
model = xgb.XGBClassifier(objective='binary:logistic')


clf = GridSearchCV(estimator = model, 
                        param_grid = parameters, 
                        scoring='precision',   #precision, recall
                        verbose=2,
                        cv=5)
```

```python
print("model tuning started")
tic = time.time()
clf.fit(X_train, y_train)
toc = time.time()


print("time(sec)) taken to tune: ",toc-tic )

print("\nclf.best_params_\n")
print(clf.best_params_)
print("score, ", clf.grid_scores_)
```

```python
# new model with optimal parameters
xgb_ = xgb.XGBClassifier(
    learning_rate =0.1, 
    n_estimators=1000,
    gamma=0.001,
    max_depth = 3,
    objective= 'binary:logistic', scale_pos_weight=1
)

# train set
xgb_.fit(X_train, y_train)

# Performance on train set
precision_ = precision_score(y_train, xgb_.predict(X_train))
print("Performance on train : ", precision_)


# test set
precision_ = precision_score(y_test, xgb_.predict(X_test))
print("Performance on test : ", precision_)


# validation set
precision_ = precision_score(y_valid, xgb_.predict(X_valid))
print("Performance on valid : ", precision_)
```

```python
print(classification_report(y_valid, xgb_.predict(X_valid)))
```
